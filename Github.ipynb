```python
import os
import subprocess
from pathlib import Path

repo_path = Path(r"C:\Users\abc\Desktop\New Folder (2)\New folder\PROVISNAL RELATED\Artificial intelligence")
os.chdir(repo_path)

repo_name = "artificial-intelligence-provisional"

# Check GitHub CLI login
login_check = subprocess.run(
    ["gh", "auth", "status"],
    capture_output=True,
    text=True
)

print(login_check.stdout)
print(login_check.stderr)

if login_check.returncode != 0:
    print("You are not logged in to GitHub CLI.")
    print("Run this in Anaconda Prompt / CMD first:")
    print("gh auth login")
else:
    # Create private GitHub repo and push current local repo
    result = subprocess.run(
        [
            "gh", "repo", "create", repo_name,
            "--private",
            "--source", ".",
            "--remote", "origin",
            "--push"
        ],
        capture_output=True,
        text=True
    )

    print(result.stdout)
    print(result.stderr)

    if result.returncode == 0:
        print("✅ Private GitHub repository created and pushed successfully.")
    else:
        print("❌ Something went wrong. Read the message above.")
```

```python
import os
import subprocess
from pathlib import Path

repo_path = Path(r"C:\Users\abc\Desktop\New Folder (2)\New folder\PROVISNAL RELATED\Artificial intelligence")
os.chdir(repo_path)

github_url = "https://github.com/Parveen117/artificial-intelligence-provisional.git"

def run(cmd):
    result = subprocess.run(cmd, capture_output=True, text=True)
    print(f"\n$ {' '.join(cmd)}")
    if result.stdout:
        print(result.stdout)
    if result.stderr:
        print(result.stderr)
    return result

# Show current status
run(["git", "status"])
run(["git", "remote", "-v"])

# Remove wrong/old origin if present
run(["git", "remote", "remove", "origin"])

# Add the already-existing GitHub repo
run(["git", "remote", "add", "origin", github_url])

# Rename local branch to main
run(["git", "branch", "-M", "main"])

# Push local files to existing GitHub repo
run(["git", "push", "-u", "origin", "main"])
```
